```python
from pathlib import Path
from zipfile import ZipFile
import sys

kit = Path("q20_12_return_board_check")
assert not kit.exists(), "目录已经存在，请不要重复解压"

with ZipFile("q20_12_return_board_check.zip") as z:
    z.extractall(".")

sys.path.insert(0, str(kit.resolve()))
import supplement_board

output_dir = supplement_board.run_images(
    Path("srcnn_q20_12_v3_resetfix"),
    kit,
)
```

```python
import json, hashlib
from pathlib import Path

def check_kit_linux(path):
    root = Path(path).resolve()
    hashes = json.loads((root / "kit_checksums.json").read_text())
    for name, expected in hashes.items():
        file = (root / name.replace("\\", "/")).resolve()
        assert file.is_relative_to(root), "路径越界"
        assert file.is_file(), f"缺少文件：{name}"
        assert hashlib.sha256(file.read_bytes()).hexdigest() == expected, \
            f"校验失败：{name}"
    return root

supplement_board.check_kit = check_kit_linux

output_dir = supplement_board.run_images(
    Path("srcnn_q20_12_v3_resetfix"),
    kit,
)
```

```python
import shutil

# 先打包刚才的板测结果
board_zip = shutil.make_archive(
    str(output_dir), "zip", root_dir=str(output_dir)
)
print("板测结果压缩包：", board_zip)

# CPU：一次预热，三次正式测量
cpu_output_dir = supplement_board.run_cpu(
    kit, case="butterfly", repeats=3
)

cpu_zip = shutil.make_archive(
    str(cpu_output_dir), "zip", root_dir=str(cpu_output_dir)
)
print("CPU结果压缩包：", cpu_zip)
```
